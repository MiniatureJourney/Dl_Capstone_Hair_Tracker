```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display, Image

ROOT = Path(r'C:\Users\Aliasgar\Desktop\DL Capstone\DL_Capstone_HairSegmentation')
LOGS_PATH = ROOT / 'outputs' / 'training_log.csv'
EVAL_PATH = ROOT / 'outputs' / 'results' / 'evaluation_report.csv'
PLOTS_DIR = ROOT / 'outputs' / 'plots'
```

## 1. Training Log Analysis

```python
if LOGS_PATH.exists():
    df_train = pd.read_csv(LOGS_PATH)
    print("Training Progress (Last 5 epochs):")
    display(df_train.tail(5))
else:
    print("Training log not found!")
```

```python
if LOGS_PATH.exists():
    fig, axes = plt.subplots(1, 2, figsize=(15, 5))
    
    # Loss Curve
    axes[0].plot(df_train['epoch'], df_train['train_loss'], label='Train Loss', color='blue', linewidth=2)
    axes[0].plot(df_train['epoch'], df_train['val_loss'], label='Val Loss', color='red', linewidth=2)
    axes[0].set_title('BCE + Dice Loss over Epochs', fontsize=14)
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].legend()
    axes[0].grid(True, linestyle='--', alpha=0.6)
    
    # Dice Curve
    axes[1].plot(df_train['epoch'], df_train['val_dice'], label='Val Dice', color='green', linewidth=2)
    axes[1].plot(df_train['epoch'], df_train['val_iou'], label='Val IoU', color='purple', linewidth=2)
    axes[1].set_title('Validation Metrics over Epochs', fontsize=14)
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Score')
    axes[1].legend()
    axes[1].grid(True, linestyle='--', alpha=0.6)
    
    plt.tight_layout()
    plt.show()
```

## 2. Final Test Set Metrics

```python
if EVAL_PATH.exists():
    df_eval = pd.read_csv(EVAL_PATH)
    df_eval.set_index('Metric', inplace=True)
    print("\n==============================")
    print("   FINAL TEST SET METRICS     ")
    print("==============================\n")
    display(df_eval)
else:
    print("Evaluation report not found!")
```
